```python
student_name = "Ngọc"
math_score = 9
science_score = 6

print(f"Tên: {student_name}, Điểm toán: {math_score}, Điểm khoa học: {science_score} ")
```

## Tính điểm

```python
a = math_score + science_score
b = math_score - science_score
c = math_score * science_score
d = math_score / science_score
print(f"Tổng điểm: {a}, Hiệu điểm: {b}, Tích điểm: {c}, Thương điểm: {d}")
```

```python
# cập nhật điểm
update = input("Nhập 1 để cập nhật điểm toán, nhập 2 để cập nhật điểm khoa học: ")

if update == '1':
    math_score1 = int(input("Nhập điểm toán được cộng: "))
    math_score += math_score1
    print(f"Điểm toán mới: {math_score}")
elif update == '2':
    science_score1 = int(input("Nhập điểm khoa học được cộng: "))
    science_score += science_score1
    print(f"Điểm khoa học mới: {science_score}")

```

```python
# đánh giá

score_total = (math_score + science_score) / 2
if score_total >= 8:
    print("Xuất sắc")
elif score_total >= 6.5:
    print("Khá")
elif score_total >= 5:
    print("Trung bình")
else:
    print("Cần cố gắng")
```
