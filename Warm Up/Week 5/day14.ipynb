```python
upload = files.upload()
img_name = list(upload.keys())[0]
img = cv2.imread(img_name)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(cv2.resize(img, (300, 300)))
```

```python
cv2.rectangle(img, (250, 100), (370, 200), (95, 95, 93), 2)
```

```python
cv2.line(img, (250, 100), (370, 200), (255, 0, 50), 3)
```

```python
# Bài 4: Làm Mờ và Phát Hiện Cạnh
# Đề bài: Viết chương trình tải ảnh lên, làm mờ ảnh bằng Gaussian Blur, phát hiện cạnh
# bằng Canny, và hiển thị cả ba ảnh (gốc, mờ, cạnh).
```

```python
blurred = cv2.GaussianBlur(img, (5, 5), 0)
blurred
```

```python
edges = cv2.Canny(blurred, 100, 200)
edges
```
