```python
# Mảng nhiệt độ
temps = np.array([25, 28, 30, 22, 35, 18, 20])
# Danh sách điều kiện
conditions = [
    temps < 20,                    
    (temps >= 20) & (temps < 30),  
    temps >= 30                   
]

categories = ["Cold", "Mild", "Hot"]

classified_temps = np.select(conditions, categories, default="Unk")

print(classified_temps)
```

```python
mat = np . random . randint (1 ,100 , (5 ,5) )
mat.shape
```

```python
mat
```

```python
# Chuẩn hóa từng hàng (trừ đi min, chia cho max - min)
mean = np.mean(mat, axis=1)
max_value = np.max(mat, axis=1)
min_value = np.min(mat, axis=1)
```

```python
norm = (mat - mean) / (max_value - min_value)
```

```python
norm
```

```python
# Áp dụng hàm softmax theo cột:

one = np.ones(5)
sigma = one @ norm
softmax = norm / sigma
```

```python
softmax
```

```python
data = np . array ([
    [1, 2, np.nan] ,
    [4, np.nan, 6] ,
    [7, 8, 9]
])
np.isnan(data)
```

```python
# Thay các giá trị NaN bằng trung bình cột
new_data = np.where(np.isnan(data), np.nanmean(data, axis=0), data)
```

```python
new_data
```

```python
# Tính z-score cho mỗi cột
mean = np.mean(new_data, axis=0)
var = (np.var(new_data, axis=0))
Z_score = (new_data - mean) / var
```

```python
Z_score
```

```python
# Chuẩn hóa min-max theo cột
min = np.min(new_data, axis=0)
max = np.max(new_data, axis=0)

Min_max_scaling = (new_data - min) / (max - min)
Min_max_scaling
```
