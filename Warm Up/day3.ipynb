```python
import math
```

```python
# hàm sigmoid
def sigmoid(x):
    return 1 / (1 + math.exp(-x))
# hàm tanh
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))

# hàm Softplus
def softplus(x):
    return math.log(1 + math.exp(x))
# hàm ReLu
def relu(x):
    return max(0, x)
# hàm LeakyReLU
def leaky_relu(x, alpha=0.01):
    return max(alpha * x, x)
# hàm PReLu
def prelu(x, alpha):
    return max(alpha * x, x) if alpha else 'alpha is not defined'
# hàm Swish
def swish(x):
    return x*sigmoid(x)
```

```python
def activate( function_name , x , alpha = None ) :
    if function_name == " ":
        return x
    elif function_name == "sigmoid" :
        return sigmoid(x)
    elif function_name == "relu" :
        return relu(x)
    elif function_name == "tanh" :
        return tanh(x)
    elif function_name == "softplus" :
        return softplus(x)
    elif function_name == "leaky_relu" :
        return leaky_relu(x, alpha) if alpha else leaky_relu(x)
    elif function_name == "prelu" :
        return prelu(x, alpha) 
    elif function_name == "swish" :
        return swish(x) 
    else :
        raise ValueError ( " Unknown activation function " )
```

```python
activation = 'relu'
a = activate(activation, 2)
print(a)
```

```python
activation = 'sigmoid'
b = activate(activation, 0)
print(b)
```

```python
activation = 'leaky_relu'
c = activate(activation, 1)
print(c)
```

```python
activation = 'prelu'
c = activate(activation, 1)
print(c)
```

```python
activation = 'prelu'
c = activate(activation, 1, 0.1)
print(c)
```
