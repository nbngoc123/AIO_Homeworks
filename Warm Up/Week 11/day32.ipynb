```python
import torch 
import numpy as np
import math
```

```python
y1 =  lambda x: x**2 + 3*x + 1
y2 =  lambda x: torch.sin(x) + torch.exp(x)
```

```python
dy_dx1 = lambda x: 2*x + 3
dy_dx2 = lambda x: torch.cos(x) + torch.exp(x)
```

```python
x = torch.tensor(3.0, requires_grad=True)

y1 = y1(x)
dy_dx1 = dy_dx1(x)
y2 = y2(x)
dy_dx2 = dy_dx2(x)
y1.backward()
y2.backward()
print(f"y1 = {y1}")
print(f"dy_dx1 = {dy_dx1}")
print(f"y2 = {y2}")
print(f"dy_dx2 = {dy_dx2}")
print(f"x.grad = {x.grad}")
```
