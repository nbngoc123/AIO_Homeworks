```python
input = torch.tensor([[1, 2], [3, 4]])
index = torch.tensor([[0, 0], [1, 0]])
result = torch.gather(input, dim=1, index=index)

print(result)
```

```python
batch1 = torch.tensor([[[1, 2], [3, 4]]])    # shape [1, 2, 2]
batch2 = torch.tensor([[[5, 6], [7, 8]]])    # shape [1, 2, 2]

output = torch.bmm(batch1, batch2)

print(output)
```

```python
batch1 = torch.tensor([[1, 2], [3, 4]])    
batch2 = torch.tensor([[5, 6], [7, 8]])    
print(torch.matmul(batch1, batch2))
```
