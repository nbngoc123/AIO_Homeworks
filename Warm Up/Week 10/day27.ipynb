```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-10, 10, 100)
y = np.sin(x)
fig, ax = plt.subplots()

ax.plot(x, y, label='sin(x)', color='blue', linestyle='--')  
ax.set_title('ham sin(x)')
ax.set_xlabel('X-axis')   # Set the x-axis label
ax.set_ylabel('Y-axis')   # Set the y-axis label

ax.grid(True)
ax.legend()
```
