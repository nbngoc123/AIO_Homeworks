```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
plt.figure(figsize=(8, 6))
ax = sns.histplot(data=tips, x='total_bill', kde=True, bins=20)
ax.set_title('phan bo total bill')
ax.set_xlabel('Total bill')
plt.show()
```

```python
sns.set_style('whitegrid')
ax = sns.violinplot(data=tips, x='time', y='total_bill', hue='sex')
ax.set_title('phan bo total bill theo time')
plt.show()
```

```python
diamonds = sns.load_dataset('diamonds')
diamonds.head()
```

```python
sns.set_style('whitegrid')               
sns.lmplot(data=diamonds, x='carat', y='price', hue='cut', fit_reg=True, scatter=True)
plt.title('phan bo price theo carat')
plt.xlabel('carat')
plt.ylabel('price')
plt.show()
```
