```python
# Bài 1: Kiểm tra phần tử trùng lặp
# Mô tả đề: Cho một mảng số nguyên nums, hãy xác định xem mảng có chứa bất kỳ phần tử nào xuất hiện nhiều hơn một lần hay không. Trả về True nếu tồn tại ít nhất một phần tử trùng lặp, và False nếu tất cả các phần tử đều khác nhau. Ví dụ Input/Output:
# • Ví dụ 1: Input: nums = [1, 2, 3, 1] Output: True Giải thích: Số 1 xuất hiện hai lần trong mảng.
# • Ví dụ 2: Input: nums = [1, 2, 3, 4] Output: False Giải thích: Tất cả các số trong mảng đều khác nhau.
# • Ví dụ 3: Input: nums = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2] Output: True Giải thích: Các số 1, 3, 4, 2 xuất hiện nhiều hơn một lần.
# Gợi ý làm bài: Sử dụng tập hợp (set) để loại bỏ các phần tử trùng lặp và so sánh độ dài của tập hợp với mảng ban đầu. Nếu độ dài khác nhau, mảng có phần tử trùng lặp. Ngoài ra, có thể dùng từ điển (dict) để đếm số lần xuất hiện của mỗi phần tử.
# 1 # C á ch 1: List + dict
# 2 def containsDuplicate ( nums ) :
# 3 # your code here
# 4 return _
# 5
# 6 # C á ch 2: List + set
# 7 def containsDuplicate ( nums ) :
# 8 # Your code here
# 9 return 
```

```python
def containsDuplicate( nums ):
    if len(nums) == len(set(nums)):
        return False
    else:
        return True
    
nums = [1, 1, 2, 3, 3, 4, 5, 6, 7, 7]
nums1 = [1, 2, 3, 4]    
containsDuplicate(nums)
```

```python
# Bài 2: Nhóm các chuỗi đảo chữ
# Mô tả đề: Cho một mảng các chuỗi strs, hãy gom các chuỗi là đảo chữ của nhau 
# (các chuỗi có cùng tập hợp ký tự với số lần xuất hiện như nhau) thành từng nhóm. 
# Trả về danh sách các nhóm, trong đó mỗi nhóm là một danh sách các chuỗi đảo chữ. 
# Thứ tự của các nhóm và chuỗi trong nhóm không quan trọng. Ví dụ Input/Output:
#  Ví dụ 1: Input: strs = [”eat”, ”tea”, ”tan”, ”ate”, ”nat”, ”bat”] Output: [[”bat”], [”nat”, ”tan”], [”ate”, ”eat”, ”tea”]] Giải thích: ”bat” là một nhóm riêng; ”nat” và ”tan” có cùng ký tự (n, a, t); ”ate”, ”eat”, ”tea” có cùng ký tự (e, a, t).
# • Ví dụ 2: Input: strs = [””] Output: [[””]] Giải thích: Chuỗi rỗng tự tạo thành một nhóm.
# • Ví dụ 3: Input: strs = [”a”] Output: [[”a”]] Giải thích: Chuỗi ”a” tự tạo thành một nhóm.
# Gợi ý làm bài: Sắp xếp từng chuỗi theo thứ tự bảng chữ cái để tạo "dấu ấn" duy nhất cho các đảo chữ, sau đó dùng từ điển (dict) để nhóm các chuỗi có cùng dấu ấn. Mỗi khóa trong từ điển là một chuỗi đã sắp xếp, và giá trị là danh sách các chuỗi tương ứng.
# 1 def groupAnagrams ( strs ) :
# 2 # Your code here
# 3 return
```

```python
def groupAnagrams(strs):
    anagram_groups = {}
    for s in strs:
        # Sắp xếp các ký tự trong chuỗi để tạo "dấu ấn"
        sorted_s = "".join(sorted(s))
        # Nếu dấu ấn đã tồn tại, thêm chuỗi vào nhóm tương ứng
        if sorted_s in anagram_groups:
            anagram_groups[sorted_s].append(s)
        # Nếu không, tạo một nhóm mới với dấu ấn này
        else:
            anagram_groups[sorted_s] = [s]
    # Chuyển đổi các nhóm thành một danh sách các danh sách
    return list(anagram_groups.values())


strs1 = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(f"Input: {strs1}, Output: {groupAnagrams(strs1)}") 

strs2 = [""]
print(f"Input: {strs2}, Output: {groupAnagrams(strs2)}") 

strs3 = ["a"]
print(f"Input: {strs3}, Output: {groupAnagrams(strs3)}")  

strs4 = ["listen", "silent", "enlist"]
print(f"Input: {strs4}, Output: {groupAnagrams(strs4)}")

strs5 = ["abc", "bac", "cab", "def", "fed"]
print(f"Input: {strs5}, Output: {groupAnagrams(strs5)}")
```

```python
# # Bài 3: Tìm cặp số có tổng bằng đích
# # Mô tả đề: Cho một mảng số nguyên nums và một số nguyên target, 
# # hãy tìm và trả về chỉ số của hai phần tử trong mảng sao cho tổng của chúng bằng 
# # target. Giả định rằng mỗi trường hợp chỉ có duy nhất một cặp giải pháp, và không 
# # được sử dụng cùng một phần tử hai lần. Thứ tự các chỉ số trong kết quả không quan trọng. Ví dụ 
# # Input/Output:
# # • Ví dụ 1: Input: nums = [2, 7, 11, 15], target = 9 Output: [0, 1] Giải thích: nums[0] + nums[1] = 2 + 7 = 9.
# # • Ví dụ 2: Input: nums = [3, 2, 4], target = 6 Output: [1, 2] Giải thích: nums[1] + nums[2] = 2 + 4 = 6.
# # • Ví dụ 3: Input: nums = [3, 3], target = 6 Output: [0, 1] Giải thích: nums[0] + nums[1] = 3 + 3 = 6.
# Gợi ý làm bài: Sử dụng từ điển (dict) để lưu giá trị đã duyệt qua cùng chỉ số của
# chúng. Với mỗi phần tử nums[i], kiểm tra xem target − nums[i] có trong từ điển hay không.
# Nếu có, trả về chỉ số của nó cùng i. Nếu không, thêm nums[i] và chỉ số i vào từ điển.
# 1 def twoSum ( nums , target ) :
# 2 # Your code here
# 3 return
```

```python
def twoSum(nums, target):

    num_map = {}
    for index, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], index]
        num_map[num] = index


nums1 = [2, 7, 11, 15]
target1 = 9
print(f"Input: nums = {nums1}, target = {target1}, Output: {twoSum(nums1, target1)}")  

nums2 = [3, 2, 4]
target2 = 6
print(f"Input: nums = {nums2}, target = {target2}, Output: {twoSum(nums2, target2)}")  

nums3 = [3, 3]
target3 = 6
print(f"Input: nums = {nums3}, target = {target3}, Output: {twoSum(nums3, target3)}")  
```
