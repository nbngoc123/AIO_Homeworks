```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 100
x = np.linspace(0, 10, n)
print(x.shape)
epsilon = np.random.normal(0, 2, n)
y = 2 * x + 1 + epsilon
print(y.shape)
trainExample = list(zip(x, y))
plt.scatter(x, y)
plt.show()
```

```python
x_expand = np.expand_dims(x, axis=1)
one = np.ones((100, 1))
x_expand.shape
```

```python
theta = lambda x: np.concatenate((x, one), axis=1)
theta(x_expand)
```

```python
def loss_function(h, y):
    return np.mean((h - y) ** 2)
```

```python
w0 = np.array([[0], 
               [0]])
l = 0.01
for i in range(50):
    h = theta(x_expand) @ w0
    GD = loss_function(h, y)
    w0 = w0 - l * 2 * theta(x_expand).T @ (h - y) 
```
