```python
# 1. Tính toán thủ công:
# • Chọn x0 = 10 và η = 0.1. Tính toán bằng tay 3 bước lặp đầu tiên của GD cho
# f(x) = x
# 2
# . Ghi lại giá trị x1, x2, x3.
# • Tính giá trị hàm mục tiêu f(xt) tại các điểm đó.
```

```python
x0 = 10 
l = 0.1  # η
# f(x) = x**2
# dX = 2*x

x1 = 10 - l*2*10 = 10 - 2 = 8  # x1 = 8
x2 = 8 - l*2*8 = 8 - 1.6 = 6.4  # x2 = 6.4
x3 = 6.4 - l*2*6.4 = 6.4 - 1.28 = 5.12  # x3 = 5.12
# f(x) = x**2
# f(x1) = 8**2 = 64
# f(x2) = 6.4**2 = 40.96
# f(x3) = 5.12**2 = 26.2144
# f(x0) = 10**2 = 100
# f(x1) < f(x2) < f(x3) < f(x0)
```

```python
# 2. Lập trình:
# • Viết chương trình bằng Python thực hiện GD cho f(x) = x
# 2
# .
# • Khởi tạo x0 = 10, η = 0.1, và lặp 100 lần hoặc cho đến khi |xt+1 − xt
# | < 10−6
# .
# • In ra giá trị xt và f(xt) sau mỗi 10 lần lặp.
# • Vẽ đồ thị của f(x) = x
# 2 và đánh dấu các điểm xt qua các lần lặp (phần vẽ đồ thị
# có thể tham khảo solution vì chưa học tới matplotlib).
```

```python

def f(x):  # f(x) = x**2
    return x**2

def df(x):  # dX = 2*x  # đạo hàm của f(x) = x**2  # dX = 2*x 
    return 2*x

def gradient_descent(x0=10, l=0.1):

    for i in range(1000):
        x_last = x0        # x bước trước đó
        x0 = x0 - l*df(x0) # x hiện tại
        if i % 10 == 0:
            print(f"x{i+1} = {x0}, f(x{i+1}) = {f(x0)}")
        if abs(x0 - x_last) < 10e-7:  # |xt+1 − xt| < 10−6  
            break
```

```python
gradient_descent()
```

```python
# 3. Thử nghiệm với η:
# • Chạy lại chương trình với η = 0.01 và η = 0.5. So sánh tốc độ hội tụ (số lần lặp
# để |xt
# | < 0.01) và rút ra kinh nghiệm về chọn giá trị này trong quá trình training.
```

```python
gradient_descent(l=0.01)
# học quá chậm và không hội tụ
```

```python

gradient_descent(l=0.3)
# học quá nhanh và không hội tụ
```
