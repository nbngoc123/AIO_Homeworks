```python
# Bài 1: Tính giai thừa bằng vòng lặp for
# Yêu cầu: Viết hàm factorial(n) sử dụng vòng lặp for để tính giai thừa của một số nguyên
# dương n. In kết quả cho n = 6.
```

```python
def factorial(n):
    result = 1
    if n < 0:
        raise ValueError("Không tính được giai thừa của số âm")
    for i in range(1, n + 1):
        result *= i
    return result
# In kết quả cho n = 6
n = 6
print(f"Giai thừa của {n} là {factorial(n)}")
```

```python
# Bài 2: Tính số e bằng chuỗi Taylor
# Yêu cầu: Viết hàm calculate_e(n) sử dụng vòng lặp for để tính gần đúng số e với n số
# hạng đầu tiên của chuỗi Taylor.
# • Tính e với n = 8.
# • So sánh với math.e.
```

```python
import math
def calculate_e(n):
    if n < 0:
        raise ValueError("n must be positive integer")
    e_approx = 1.0
    for i in range(1, n):
        e_approx += 1 / factorial(i)

    return e_approx

def math_e(n):
    return math.exp(n)
# Tính e với n = 8
n = 8
e_approx = calculate_e(n)
print(f"Giá trị gần đúng của e với {n} số hạng đầu tiên của chuỗi Taylor là {e_approx}")
print(f"Giá trị thực của e là {math.e}")
print(f"Sự chênh lệch là {abs(e_approx - math.e)}")
```

```python
# Bài 3: Tính ex bằng chuỗi Taylor
# Yêu cầu: Viết hàm calculate_exp(x, n) sử dụng vòng lặp for để tính gần đúng ex với n
# số hạng đầu tiên của chuỗi Taylor.
# • Tính e2 với n = 10.
# • So sánh với giá trị thực từ math.exp(2).
```

```python
def calcuulate_exp(x, n):
    exp_approx = 1.0
    if n > 0: 
        for i in range(1, n):
            exp_approx += (x ** i) / factorial(i)
    else:
        raise ValueError("n must be positive integer")
    return exp_approx

x = 2
n = 10
exp_approx = calcuulate_exp(x, n)
print(f"Giá trị gần đúng của e^{x} với {n} số hạng đầu tiên của chuỗi Taylor là {exp_approx}")
print(f"Giá trị thực của e^{x} là {math.exp(x)}")
print(f"Sự chênh lệch là {abs(exp_approx - math.exp(x))}")
```
