```python
# Bài tập 1: Ứng dụng vòng lặp While
# Đề bài: Viết chương trình sử dụng vòng lặp while yêu cầu người dùng nhập vào một số
# nguyên dương n, sau đó tính tổng các chữ số của n.
# Gợi ý:
# • Nhập số nguyên dương n.
# • Khởi tạo tổng ban đầu S = 0.
# • Sử dụng vòng lặp while để lấy từng chữ số cuối cùng bằng phép chia dư.
# • Cộng chữ số vào S, sau đó loại bỏ chữ số cuối bằng phép chia nguyên.
# • Lặp lại cho đến khi n = 0, sau đó in kết quả.
```

```python
def get_sum(n):
    if not isinstance(n, int) or n < 0:
        raise TypeError("n must be an integer")
    s = 0
    while n // 10:
        s = s + n % 10
        n = n // 10
    print(s + n)
```

```python
get_sum(123)
```

```python
# Bài tập 2: Dùng vòng lặp While để giải phương trình tìm nghiệm
# bằng phương pháp Newton
# Đề bài: Viết chương trình sử dụng vòng lặp while để tìm nghiệm của phương trình f (x) =
# x3 − 3x + 1 = 0 bằng phương pháp Newton. Yêu cầu:
# • Chọn giá trị ban đầu x0 = 5.
# Dừng khi sai số giữa hai lần lặp liên tiếp nhỏ hơn ϵ = 0.0001.
# • In ra giá trị nghiệm gần đúng và số lần lặp.
# Gợi ý:
# • Hàm f (x) = x3 − 3x + 1 và đạo hàm f ′(x) = 3x2 − 3 được định nghĩa.
# • Vòng lặp while True chạy vô hạn, nhưng sử dụng break để thoát khi:
# – Sai số |xn+1 − xn| < ϵ.
# – Hoặc vượt quá số lần lặp tối đa (tránh vòng lặp vô hạn).
# • Mỗi lần lặp, công thức Newton được áp dụng: xn+1 = xn − f (xn)
# f ′(xn) .
# • Kết quả là nghiệm gần đúng của phương trình.
```

```python
def find_x(x=100):
    threshold = 0.00001
    while True:
        f_x = x**3 - 3*x + 1
        df_dx = 3*(x**2) - 3

        x = x - (f_x / df_dx)
        print(x) # theo dõi quá trình tìm nghiệm
        if abs(f_x / df_dx) < threshold:
            break
    return x, f_x
```

```python
x, fx = find_x()
```
