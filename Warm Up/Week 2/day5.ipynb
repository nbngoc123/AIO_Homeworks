```python
# Bài tập 1: Tìm tất cả các số nguyên tố trong khoảng từ 1 đến N
# Mô tả bài toán
# Viết một chương trình yêu cầu người dùng nhập một số nguyên dương N , sau đó in ra tất
# cả các số nguyên tố từ 1 đến N . Một số nguyên tố là số chỉ chia hết cho 1 và chính nó (ví
# dụ: 2, 3, 5, 7,...).
# Mô tả Input, Output
# • Input: Một số nguyên dương N (do người dùng nhập).
# • Output: Các số nguyên tố từ 1 đến N , cách nhau bởi dấu phẩy và khoảng trắng (",
# ").
# • Ví dụ:
# – Input: 10
# – Output: 2, 3, 5, 7
```

```python
import math

def is_prime(num):
        if not isinstance(num, int) or num < 0:
            raise TypeError("N phải là số nguyên dương")
        if num <= 1:
            return False    
        for i in range(2, int(math.sqrt(num)) + 1):
            if num % i == 0:
                return False
        return True

def find_primes(N):
    for i in range(N + 1):
        if is_prime(i) == True:
            print(i, end=", ")
        
```

```python
find_primes(10)
```

```python
# Bài tập 2: Tính tổng các số hoàn hảo nhỏ hơn N
# Mô tả bài toán
# Viết một chương trình yêu cầu người dùng nhập một số nguyên dương N , sau đó tính tổng
# tất cả các số hoàn hảo nhỏ hơn N . Một số hoàn hảo là số mà tổng các ước số thực sự của
# nó (không tính chính nó) bằng chính số đó (ví dụ: 6 = 1 + 2 + 3).
# Mô tả Input, Output
# • Input: Một số nguyên dương N (do người dùng nhập).
# • Output: Tổng của các số hoàn hảo nhỏ hơn N .
# • Ví dụ:
# – Input: 10
# – Output: 6 (vì 6 là số hoàn hảo duy nhất nhỏ hơn 10)
```

```python
def sum_perfect_numbers(N):
    sum = 0
    for i in range(1, N):
        if is_perfect(i) == True:
            sum += i
    return sum
def is_perfect(num):
    sum = 0
    for i in range(1, num):
        if num % i == 0:
            sum += i
    if sum == num:
        return True
    else:
        return False
    

print(sum_perfect_numbers(30))
```
