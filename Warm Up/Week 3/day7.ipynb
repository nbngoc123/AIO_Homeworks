```python
# Bài 1: Tạo danh sách số nguyên từ bàn phím
# Yêu cầu:
# 1. Nhập số lượng n từ bàn phím (số phần tử của List).
# 2. Nhập n số nguyên từ bàn phím và lưu vào List.
# 3. In ra List vừa tạo.
# Ví dụ:
# Nhập số lượng phần tử: 4
# Nhập số thứ 1: 5
# Nhập số thứ 2: 2
# Nhập số thứ 3: 8
# Nhập số thứ 4: 1
# List: [5, 2, 8, 1]
```

```python
def print_array():
    array = []
    n = int(input('số phần tử mà bạn muốn lưu: '))
    if (not isinstance(n, int) or n < 0):
        raise TypeError("bắt buộc phải là số nguyên dương")
    
    for i in range(n):
        number = int(input(f"số thứ {i} là: "))
        array.append(number)

    print(array)
        
```

```python
print_array()
```

```python
# Bài 2: Tính toán trên List
# Yêu cầu:
# 1. Dùng List từ Bài 1 (hoặc tạo List cố định: [5, 2, 8, 1]).
# 2. Tính tổng các phần tử bằng sum().
# 3. Tìm giá trị lớn nhất và nhỏ nhất bằng max() và min().
# 4. In kết quả.
# Ví dụ::
# List: [5, 2, 8, 1]
# Tổng: 16
# Giá trị lớn nhất: 8
# Giá trị nhỏ nhất: 1
```

```python
# # Bài 3: Thêm và xóa phần tử
# # Yêu cầu:
# # 1. Tạo List ban đầu: numbers = [10, 20, 30, 40].
# # 2. Thêm một phần tử (nhập từ bàn phím) vào cuối List bằng append().
# # 3. Chèn một phần tử (nhập từ bàn phím) vào vị trí bất kỳ (nhập chỉ số từ bàn phím)
# # bằng insert().
# # 4. Xóa một phần tử theo giá trị (nhập từ bàn phím) bằng remove().
# # 5. In List sau mỗi thao tác.
# Ví dụ::
# List ban đầu: [10, 20, 30, 40]
# Nhập phần tử để thêm vào cuối: 50
# List sau khi thêm: [10, 20, 30, 40, 50]
# Nhập vị trí để chèn: 2
# Nhập phần tử để chèn: 25
# List sau khi chèn: [10, 20, 25, 30, 40, 50]
# Nhập giá trị để xóa: 20
# List sau khi xóa: [10, 25, 30, 40, 50]
```

```python
array = [1, 2, 3, 4, 5]
cac_chuc_nang = ['Nhập phần tử để thêm vào cuối', 'Nhập vị trí để chèn', 'Nhập giá trị để xóa']
ket_qua = ['List sau khi thêm', 'List sau khi chèn', 'List sau khi xóa']

for chuc_nang in cac_chuc_nang:
    value_input = input(f'{chuc_nang}: ')

    if chuc_nang == cac_chuc_nang[0]:
        array.append(int(value_input))
    elif chuc_nang == cac_chuc_nang[1]:
        value_insert = int(input('giá trị để chèn là: '))
        position = int(value_input)
        array.insert(position, value_insert)
    else:
        (array.remove(int(value_input)))

    print(f'{ket_qua[cac_chuc_nang.index(chuc_nang)]}: {array}')
    
```

```python
# Bài 4: Duyệt List bằng vòng lặp
# Yêu cầu:
# 1. Dùng List từ Bài 3 (hoặc tạo List cố định: [10, 25, 30, 40, 50]).
# 2. Dùng vòng lặp for để in từng phần tử kèm chỉ số index của nó.
# 3. Dùng vòng lặp while để in lại tương tự.
# Ví dụ::
# Duyệt bằng for:
# Phần tử tại chỉ số 0 là 10
# Phần tử tại chỉ số 1 là 25
# Phần tử tại chỉ số 2 là 30
# Phần tử tại chỉ số 3 là 40
# Phần tử tại chỉ số 4 là 50
# Duyệt bằng while:
# Phần tử tại chỉ số 0 là 10
# Phần tử tại chỉ số 1 là 25
# Phần tử tại chỉ số 2 là 30
# Phần tử tại chỉ số 3 là 40
# Phần tử tại chỉ số 4 là 50
```

```python
array = [10, 20, 30, 40, 50]
for i in range(len(array)):
    print(f'Phần tử tại chỉ số {i} là {array[i]}')
chi_so = 0
print()
while (chi_so < len(array)):
    print(f'Phần tử tại chỉ số {chi_so} là {array[chi_so]}')
    chi_so += 1
```

```python
# Bài 5: Tìm kiếm phần tử trong List
# Yêu cầu:
# 1. Dùng List từ Bài 4 (hoặc tạo List cố định: [10, 25, 30, 40, 50]).
# 2. Nhập một số từ bàn phím.
# 3. Kiểm tra số đó có trong List không (dùng in).
# 4. Nếu có, in ra vị trí đầu tiên của nó bằng index(). Nếu không, thông báo "Không tìm
# thấy".
# Ví dụ::
# List: [10, 25, 30, 40, 50]
# Nhập số cần tìm: 30
# 30 có trong List, vị trí: 2
# Nhập số cần tìm: 15
# Không tìm thấy 15 trong List

```

```python
array = [10, 25, 30, 40, 50]
def bai_5():
    print(f'list: {array}')
    n = input(f'Nhập số cần tìm: ')
    if int(n) in array:
        print(f'{n} có trong danh sách, vị trí là {array.index(int(n))}')
    else:
        print(f'{n} không có trong danh sách')
```
