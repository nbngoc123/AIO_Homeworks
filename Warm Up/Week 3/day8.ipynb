```python
# Bài 2: Copy List & Aliasing
# Yêu cầu:
# 1. Tạo List ban đầu: original = [1, 2, [3, 4]].
# 2. Tạo 2 bản sao: một bằng copy() và một bằng deepcopy().
# 3. Thay đổi phần tử lồng nhau trong cả hai bản sao (VD: đổi [3, 4] thành [10, 4]) và
# in cả 3 List để so sánh.
# Ví dụ:
# Original ban đầu: [1, 2, [3, 4]]
# Shallow copy sau khi sửa: [1, 2, [10, 4]]
# Deep copy sau khi sửa: [1, 2, [10, 4]]
# Original sau khi sửa: [1, 2, [10, 4]]
```

```python
import copy
list_1 = [1, 2, [3, 4]]
copy_list = list_1.copy()
deepcopy_list = copy.deepcopy(list_1)

copy_list[2] = [10, 4]
print(copy_list)
deepcopy_list[2] = [10, 4]
print(deepcopy_list)
list_1[2] = [10, 4]
print(list_1)
```

```python
# Dictionary
# Yêu cầu:
# 1. Tạo Dictionary: students = "An": 8.5, "Bình": 7.0.
# 2. Nhập tên và điểm của một học sinh mới từ bàn phím, thêm vào Dictionary.
# 3. In tất cả tên học sinh (dùng .keys()) và điểm trung bình (dùng .values()).
# Ví dụ:
# Dictionary ban đầu: ’An’: 8.5, ’Bình’: 7.0
# Nhập tên học sinh: Cường
# Nhập điểm: 9.0
# Dictionary sau khi thêm: ’An’: 8.5, ’Bình’: 7.0, ’Cường’: 9.0
# Danh sách học sinh: dict_keys([’An’, ’Bình’, ’Cường’])
# Điểm trung bình: 8.17
```

```python
students = {
    "An": 3.9,
    "Ngọc": 9.9,
}
def render(dic):
    for key, value in dic.items():
        print(f"{key}: {value}", end=", ")
    print()

print(f'Dictionary ban đầu: ', end='')
render(students)
name = input("Tên: ")
score = float(input("Điểm: "))
students[name] = score
print(f'Dictionary sau khi thêm: ', end='')
render(students)
print(f'Danh sách hs: {list(students.keys())}')
print(f'Điểm tb: {sum(students.values()) / len(students)}')
```
